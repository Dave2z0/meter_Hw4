```python
from ann_package.ann_model import ANNModel
import itertools
import numpy as np
```

```python
hidden_layer_sizes_options = [(50,), (100,), (200,), (100, 50)]
learning_rates = [0.01, 0.001, 0.0001]
max_iterations = [200, 500, 1000]
```

```python
# Generate all combinations
param_combinations = list(itertools.product(hidden_layer_sizes_options, learning_rates, max_iterations))

# Store results
results = []

# Loop through each combination
for hidden_layers, lr, max_iter in param_combinations:
    print(f"\nTesting {hidden_layers}, Learning Rate: {lr}, Max Iter: {max_iter}")
    
    # Create ANN model instance
    ann = ANNModel("data/Meter_A.txt", hidden_layer_sizes=hidden_layers, learning_rate_init=lr, max_iter=max_iter)
    
    # Load, preprocess, train, and test
    ann.load_and_preprocess_data()
    ann.train_model()
    accuracy = ann.test_model()

    # Store results
    results.append({
        "hidden_layers": hidden_layers,
        "learning_rate": lr,
        "max_iter": max_iter,
        "accuracy": accuracy
    })

# Convert results to a DataFrame for better visualization
import pandas as pd
df_results = pd.DataFrame(results)

# Save results to CSV file
df_results.to_csv("hyperparameter_tuning_results.csv", index=False)

# Print results
print("\nHyperparameter Tuning Results:")
print(df_results)
```

# **Hyperparameter Tuning Report**

## **1. Summary of Results**
- The **highest accuracy achieved was 94.44%** with:
  - **Hidden Layers:** `(200,)`
  - **Learning Rate:** `0.0001`
  - **Max Iterations:** `200`
  
- The **second-best accuracy (88.88%)** was achieved with:
  - **Hidden Layers:** `(50,)` and `(100,)`
  - **Learning Rate:** `0.001` or `0.0001`
  - **Max Iterations:** `200`, `500`, or `1000`

- Most other configurations resulted in **accuracy around 83.33%**.

---

## **2. Key Observations**
- **More neurons (`200,`) improved accuracy** to 94.44% when combined with a low learning rate (`0.0001`).
- **Lower learning rates (`0.0001` and `0.001`) generally performed better** than `0.01`, showing that smaller updates helped improve learning stability.
- **More iterations (`max_iter = 1000`) helped in some cases**, but in most cases, accuracy **did not improve significantly after 500 iterations**.
- **Adding extra layers (`100, 50`) did not lead to higher accuracy**
