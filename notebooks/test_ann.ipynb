```python
from ann_package.ann_model import ANNModel
```

```python
# Define file path
file_path = "data/Meter_A.txt"  # Adjust based on dataset location

# Create ANN model instance
ann = ANNModel(file_path, target_column="target", hidden_layer_sizes=(100,), learning_rate_init=0.01, max_iter=300)

# Load and preprocess data
ann.load_data()
ann.preprocess_data(target_column="target")

# Train and test the model
ann.train_model()
accuracy = ann.test_model()

print(f"Final Accuracy: {accuracy:.4f}")
```

```python
param_combinations = [
    {"hidden_layer_sizes": (50,), "learning_rate_init": 0.01, "max_iter": 200},
    {"hidden_layer_sizes": (100,), "learning_rate_init": 0.001, "max_iter": 500},
    {"hidden_layer_sizes": (200,), "learning_rate_init": 0.0001, "max_iter": 1000},
    {"hidden_layer_sizes": (100, 50), "learning_rate_init": 0.01, "max_iter": 500},
]

for params in param_combinations:
    print(f"Testing {params}...")
    ann = ANNModel(file_path, target_column="target", **params)
    ann.load_data()
    ann.preprocess_data(target_column="target")
    ann.train_model()
    accuracy = ann.test_model()
    print(f"Results for {params}: Accuracy = {accuracy:.4f}")
```

# Observation
Best accuracy observed: 83.33%
Hyperparameter tuning did not improve accuracy.
Potential causes: Linearly separable data, limited feature representation, or overfitting.
